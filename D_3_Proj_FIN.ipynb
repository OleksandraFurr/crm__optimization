```python
ua_deals = df_deals['Contact Name'].nunique()
print('ua_deals =', ua_deals)
ua_calls = df_calls['CONTACTID'].nunique()
print('ua_calls =',ua_calls)
ua_conts = df_cont['Id'].nunique()
print('ua_conts =',ua_conts)
```

Wir nehmen für die Berechnungen den größten Wert UA = 18510.

```python
ua = ua_conts
```

```python
df_filt = df_deals[df_deals['Stage'] == 'Payment Done']
```

```python
df_ux_ui = df_filt[df_filt['Product'] == 'UX/UI Design']
df_dig_mark = df_filt[df_filt['Product'] == 'Digital Marketing']
df_web_dev = df_filt[df_filt['Product'] == 'Web Developer']
```

```python
b = len(df_filt)
b_dig_mark = len(df_dig_mark)
b_ux_ui = len(df_ux_ui)
b_web_dev = len(df_web_dev)
```

```python
c1 = b/ua_conts
c1_dig_mark = b_dig_mark/ua_conts
c1_ux_ui = b_ux_ui/ua_conts
c1_web_dev = b_web_dev/ua_conts
```

```python
df_spend.head(2)
```

```python
ac = df_spend['Spend'].sum()
```

```python
t = df_filt['Months of study'].sum()
t_dig_mark = df_dig_mark['Months of study'].sum()
t_ux_ui = df_ux_ui['Months of study'].sum()
t_web_dev = df_web_dev['Months of study'].sum()
```

```python
df_deals['Product'].value_counts()
```

```python
diff_amount = df_deals['Offer Total Amount'] - df_deals['Initial Amount Paid']
diff_course = df_deals['Course duration'] - 1
diff_mounth = df_deals['Months of study'] - 1
rest_payment = (((diff_amount / diff_course) * diff_mounth) + df_deals['Initial Amount Paid'])
```

```python
df_deals['AOV_i'] = np.where(
    (df_deals['Offer Total Amount'] - df_deals['Initial Amount Paid']) > 0,
    rest_payment / df_deals['Months of study'],
    df_deals['Offer Total Amount'] / df_deals['Course duration']
)

df_deals['R_i'] = df_deals['AOV_i'] * df_deals['Months of study']

df_ux_ui = df_deals[df_deals['Product'] == 'UX/UI Design']
df_dig_mark = df_deals[df_deals['Product'] == 'Digital Marketing']
df_web_dev = df_deals[df_deals['Product'] == 'Web Developer']
```

```python
revenu = df_deals['R_i'].sum()
revenu
```

```python
df_deals[df_deals['Months of study'] == 7]
```

```python
rev_dig_mark = df_dig_mark['R_i'].sum()
rev_ux_ui = df_ux_ui['R_i'].sum()
rev_web_dev = df_web_dev['R_i'].sum()
```

```python
aov = revenu / t
aov_dig_mark = rev_dig_mark / t_dig_mark
aov_ux_ui = rev_ux_ui / t_ux_ui
aov_web_dev = rev_web_dev / t_web_dev
```

```python
apc = t / b
apc_dig_mark = t_dig_mark / b_dig_mark
apc_ux_ui = t_ux_ui / b_ux_ui
apc_web_dev = t_web_dev / b_web_dev
```

```python
cltv = aov * apc
cltv_dig_mark = aov_dig_mark * apc_dig_mark
cltv_ux_ui = aov_ux_ui * apc_ux_ui
cltv_web_dev = aov_web_dev * apc_web_dev
```

```python
ltv = (cltv * c1)
ltv_dig_mark =(cltv_dig_mark * c1_dig_mark)
ltv_ux_ui = (cltv_ux_ui * c1_ux_ui)
ltv_web_dev = (cltv_web_dev * c1_web_dev)
```

```python
cpa = ac / ua
```

```python
cm = ua*(ltv - cpa)
cm_dig_mark = ua*(ltv_dig_mark - cpa)
cm_ux_ui = ua*(ltv_ux_ui - cpa)
cm_web_dev = ua*(ltv_web_dev - cpa)
```

```python
def add_row(df, product, **metrics):
    row = {'Product': product}
    row.update(metrics)
    df.loc[len(df)] = row
    return df
```

```python
columns = ['Product','UA', 'B', 'C1', 'REV', 'T','AOV', 'APC', 'CLTV', 'LTV', 'AC', 'CPA', 'CM']
```

```python
df1 = pd.DataFrame(columns=columns)
```

```python
df1 = add_row(df1, 'all', UA=ua, B=b, C1=c1, REV=revenu, T=t, AOV=aov,
              APC=apc, CLTV=cltv, LTV=ltv, AC=ac, CPA=cpa, CM=cm)

df1 = add_row(df1, 'Digital Marketing', UA=ua, B=b_dig_mark, C1=c1_dig_mark,
              T=t_dig_mark, REV=rev_dig_mark, AOV=aov_dig_mark,
              APC=apc_dig_mark, CLTV=cltv_dig_mark, LTV=ltv_dig_mark, AC=ac,
              CPA=cpa, CM=cm_dig_mark)

df1 = add_row(df1, 'UX Design', UA=ua, B=b_ux_ui, C1=c1_ux_ui, T=t_ux_ui,
              REV=rev_ux_ui, AOV=aov_ux_ui, APC=apc_ux_ui, CLTV=cltv_ux_ui,
              LTV=ltv_ux_ui, AC=ac, CPA=cpa, CM=cm_ux_ui)

df1 = add_row(df1, 'Web Developer', UA=ua, B=b_web_dev, C1=c1_web_dev,
              T=t_web_dev, REV=rev_web_dev, AOV=aov_web_dev,
              APC=apc_web_dev, CLTV=cltv_web_dev, LTV=ltv_web_dev, AC=ac,
              CPA=cpa, CM=cm_web_dev)
```
